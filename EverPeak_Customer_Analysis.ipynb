## Function to Fill Missing Values

### 🎯 Objective: Create a function that converts columns to numeric type and fills missing values with the mean.

```python
# Create function
def rellenar_ausentes(df, cols_fill):

    for col in cols_fill:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        med = df[col].mean()
        df[col] = df[col].fillna(med, inplace=False)

    return df

# --------------------------------------------------------------------
# Import library and load data
import pandas as pd
df = pd.read_csv("/datasets/everpeak_retail.csv")

# Check initial missing values
print("Initial missing values:")
print(df[["customer_age", "price"]].isna().sum())

# Define columns to fill
columnas_rellenar = ["customer_age", "price"]

# Apply the function and review the changes
df = rellenar_ausentes(df, columnas_rellenar)
print("\nMissing values after cleaning:")
print(df[["customer_age", "price"]].isna().sum())
```

## Build a Data-Cleaning Pipeline

### 🎯 Objective: Create a pipeline function that orchestrates all cleaning functions in the correct order.

```python
    # Helper functions (created in previous tasks)
def reemplazar_sentinels(df, sentinels, numeric_cols):
    for col in numeric_cols:
        df[col] = df[col].replace(sentinels, pd.NA)
    return df

def rellenar_ausentes(df, cols_fill):
    for col in cols_fill:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col].fillna(df[col].mean(), inplace=True)
    return df

# Create function pipeline
def limpiar_df(df):
    # define sentinel/error values
    valores_erroneos = [-999,999,0,-1]
    # define numeric columns
    columnas_numericas = ["customer_age","price"]

    df = reemplazar_sentinels(df, valores_erroneos, columnas_numericas)

    df = rellenar_ausentes(df, columnas_numericas)

    return df

# --------------------------------------------------------------------
# Import library and load data
import pandas as pd
df = pd.read_csv("/datasets/everpeak_retail.csv")

# Check initial missing values
print("Initial missing values:")
print(df[["customer_age", "price"]].isna().sum())

# Apply the complete pipeline
df = limpiar_df(df)
print("\nMissing values después del pipeline:")
print(df[["customer_age", "price"]].isna().sum())
```

## Guided Practice 1 — Detect Outliers with IQR

### 🎯 Objective: Identify outliers in `order_value` using the IQR method.

```python
import pandas as pd
df = pd.read_csv("/datasets/everpeak_clean.csv")

# Calculate Q1, Q3, and IQR
Q1  = df['order_value'].quantile(0.25)
Q3  = df['order_value'].quantile(0.75)  # Tercer cuartil
IQR = Q3 - Q1  # Segundo cuartil o mediana

# Calculate lower and upper bounds
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Display results
print('First quartile: ', Q1)
print('Third quartile: ', Q3)
print('IQR: ', IQR)

print("\nRecords below the lower bound")
print(df[df["order_value"] < limite_inferior])

print("\nRecords above the upper bound")
print(df[df["order_value"] > limite_superior])
```

## Guided Practice 2 — Z-Scores

### 🎯 Objective: Identify outliers in `order_value` using Z-Scores.

```python
import pandas as pd
df = pd.read_csv("/datasets/everpeak_clean.csv")

# Calculate the mean
mean = df["order_value"].mean()

# Calculate the standard deviation
std = df["order_value"].std()

# Create the Z-score
df['z'] = (df['order_value'] - mean) / std

# Calculate extreme values
print(df[df["z"] > 3])
```

## Individual Practice 1 — Compare Methods for `price`

### 🎯 Objective: Compare the IQR and Z-Score methods for the `price` column.

```python
import pandas as pd
df = pd.read_csv("/datasets/everpeak_clean.csv")

# ----- IQR -----
# Calculate Q1, Q3, and IQR
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# ----- Z-score -----
# Calculate mean, standard deviation, and Z-score
mean = df["price"].mean()
std = df["price"].std()
df['z'] = (df["price"] - mean) / std

# ----- Mostrar resultados -----
print('Outliers using IQR:')
print(df[(df["price"] < lower) | (df["price"] > upper)])


print('\nOutliers using Z-Score:')
print(df[abs(df["z"]) > 3])
```

## Guided Practice 2 — Advanced Segmentation with a Function and `apply`

### 🎯 Objective: The marketing management team has asked us to segment customers by age and purchase volume into four segments.

```python
import pandas as pd
df = pd.read_csv("/datasets/everpeak_clean.csv")

def classify_volume(row):
    age = row['customer_age']
    qty = row['quantity']

    # 1. Handle null/missing values
    # pd.isna() robustly checks whether a value is NaN
    if pd.isna(age) or pd.isna(qty):
        return "Data Error"

    # --- 2. Segmentación de Altas Countes ---
    if qty > 22:
        if age > 55:
            return "Sr. High Volume"
        else:
            return "Jr. High Volume"
    elif qty <= 22:
        if age > 55:
            return "Sr. Low Volume"
        else: # age < 55
            return "Jr. Low Volume"

# Apply the function and verify changes
df["volume_segment"] = df.apply(classify_volume, axis=1)
print(df['volume_segment'].value_counts())
```

## Individual Practice 1 — Advanced Segmentation with a Function and `apply`

### 🎯 Objective: The finance management team has asked us to segment transactions by payment method and purchase volume to analyze payment patterns for high- and low-volume purchases.

```python
import pandas as pd
df = pd.read_csv("/datasets/everpeak_retail.csv")

def classify_payment(row):
    card = row["payment_method"]
    qty = row["quantity"]

    # 1. Handle null/missing values
    if pd.isna(card) or pd.isna(qty):
        return "Data Error"

    if qty > 22:
        if card == "credit_card" or card == "debit_card":
            return "card_high_volume"
        else:
            return "no_card_high_volume"
    elif qty <= 22:
        if card == "credit_card" or card == "debit_card":
            return "card_low_volume"
        else:
            return "no_card_low_volume"

# Apply the function and verify changes
df['payment_segment'] = df.apply(classify_payment, axis=1)
print(df['payment_segment'].value_counts())

# New analysis: detect outliers using IQR
Q1 = df['order_value'].quantile(0.25)
Q3 = df['order_value'].quantile(0.75)
IQR = Q3 - Q1
upper = Q3 + 1.5*IQR
df['outlier_flag'] = df['order_value'] > upper

# Add a missingness indicator column
df['payment_segment_missing'] = df['payment_segment'].isna()
```
